```python
fig, ax = plt.subplots(figsize=(10, 8))

x_line = np.linspace(0.5, 3.5, 100)
y_blue = 4.0 * np.exp(-x_line)
ax.plot(x_line, y_blue, color='blue', linewidth=2, label='Синяя линия')

y_red = 0.5 * np.exp(x_line / 1.5)
ax.plot(x_line, y_red, color='red', linewidth=2, label='Красная линия')

np.random.seed(42)
x_scatter = np.random.uniform(0.5, 3.5, 40)
y_scatter = np.random.uniform(0.5, 3.5, 40)
ax.scatter(x_scatter, y_scatter, facecolors='none', edgecolors='gray', s=40)

ax.set_xlim(0, 4)
ax.set_ylim(0, 4)

ax.set_xticks(np.arange(0, 4.5, 0.5))
ax.set_yticks(np.arange(0, 4.5, 1.0))
ax.set_xticks(np.arange(0, 4.1, 0.25), minor=True)
ax.set_yticks(np.arange(0, 4.1, 0.5), minor=True)

ax.grid(which='major', color='gray', linestyle='--', linewidth=1)

ax.set_title("Элементы изображения", fontsize=16)

ax.set_xlabel("Подпись оси ОХ\n(X axis label)", fontsize=12, labelpad=10)
ax.set_ylabel("Подпись оси OY\n(Y axis label)", fontsize=12, labelpad=10)

ax.legend(loc='upper right')

ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.spines['bottom'].set_visible(True)
ax.spines['left'].set_visible(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-4, 5, 100)

# 2. Вычисляем y по формуле
y = x**2 - x - 6

# 3. Создаем график
plt.figure(figsize=(8, 5))
plt.plot(x, y, color='blue', linewidth=2, label='y = x^2 - x - 6')

# 4. Горизонатльная линия y = 0
plt.axhline(0, color='black', linewidth=1.5, linestyle=':')

# 5. Оформление
plt.title("График функции y(x) = x^2 - x - 6", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y(x)", fontsize=12)

# Сетка для наглядности
plt.grid(True, linestyle=':', alpha=0.7)

# 6. Корни точками на графике
plt.scatter([-2, 3], [0, 0], color='red', zorder=5, s=50)
plt.text(-2.2, -2, 'x = -2', color='red', fontsize=12, fontweight='bold')
plt.text(2.8, -2, 'x = 3', color='red', fontsize=12, fontweight='bold')

plt.show()
```

```python
print("График 1")

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

np.random.seed(42)
data1 = np.random.normal(1.5, 0.3, 100)
data2 = np.random.normal(4.8, 0.6, 100)
data3 = np.random.normal(3.0, 0.5, 20)

data = np.concatenate([data1, data2, data3])

plt.figure(figsize=(7, 5))

plt.hist(data, bins=7, density=True, color='pink', alpha=0.8, edgecolor=None)

kde = gaussian_kde(data)
x_vals = np.linspace(min(data) - 1, max(data) + 1, 200)
plt.plot(x_vals, kde(x_vals), color='palevioletred', linewidth=2)

plt.xlabel('petal_length', fontsize=10)
plt.ylabel('Density', fontsize=10)

plt.show()
```

```python
print("График 2")

x = np.linspace(0, 100, 100)
y = 2 * x

fig, ax = plt.subplots(figsize=(7, 5))

ax.plot(x, y, color='steelblue', linewidth=2)
ax.set_xlabel('x')
ax.set_ylabel('y')

ax_inset = fig.add_axes([0.55, 0.45, 0.25, 0.25])

ax_inset.plot(x, y, color='steelblue', linewidth=1.5)
ax_inset.set_xlabel('x', fontsize=8)
ax_inset.set_ylabel('y', fontsize=8)

ax_inset.set_xticks([0, 100])
ax_inset.set_yticks([0, 200])
ax_inset.tick_params(axis='both', labelsize=8)

for spine in ax_inset.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1)

plt.show()
```

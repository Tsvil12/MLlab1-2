# Задание 1:

Напишите программы, которые создают следующие массивы NumPy:
1. Массив со значениями 1, 7, 13, 105. Посчитайте и выведите на экран размер
памяти, который он занимает. Сохраните массив в текстовый и бинарный файлы.
Затем загрузите его и выведите на экран.
2. Три массива: из 10 нулей, 10 единиц, 10 пятерок.
3. Массив последовательности четных целых чисел от 30 до 70.
4. Массив из 10 элементов со значениями, равномерно распределенными между 5 и 50.
5. Массив 3x3x3 из 27 случайных чисел в диапазоне от 1 до 100.
6. Массив 3х4, заполненной значениями от 30 до 41.
7. Массив 10x10, в котором элементы на границах равны 1, а внутри 0.
8. Массив 5x5 из нулей с элементами на главной диагонали равными 1, 2, 3, 4, 5.
9. Массив 4x4, в котором 0 и 1 расположены в шахматном порядке, с нулями на
главной диагонали.
10. Массив, заполненный всеми днями марта 2017 года. Тип данных массива – дата.

```python
import numpy as np
import datetime as dt

print("Пункт 1: массив [1, 7, 13, 105]")
arr1 = np.array([1, 7, 13, 105])
print("Массив:", arr1)
print("Размер памяти (байт):", arr1.nbytes)
print("Размер одного элемента (байт):", arr1.itemsize)

np.savetxt("arr1.txt", arr1)
np.save("arr1.npy", arr1)

loaded_txt = np.loadtxt("arr1.txt")
loaded_bin = np.load("arr1.npy")
print("Загружено из txt:", loaded_txt)
print("Загружено из npy:", loaded_bin)

print("\nПункт 2: нули, единицы, пятёрки")
zeros = np.zeros(10, dtype=int)
ones = np.ones(10, dtype=int)
fives = np.full(10, 5, dtype=int)
print("Нули:    ", zeros)
print("Единицы: ", ones)
print("Пятёрки: ", fives)

print("\nПункт 3: чётные числа от 30 до 70")
evens = np.arange(30, 71, 2)
print(evens)

print("\nПункт 4: 10 равномерно распределённых значений от 5 до 50")
uniform_arr = np.linspace(5, 50, 10)
print(uniform_arr)

print("\nПункт 5: 3x3x3 случайных чисел в диапазоне [1, 100)")
rand_arr = np.random.randint(1, 101, size=(3, 3, 3))
print(rand_arr)

print("\nПункт 6: 3x4, значения от 30 до 41")
arr6 = np.arange(30, 42).reshape(3, 4)
print(arr6)

print("\nПункт 7: 10x10, границы = 1, внутри = 0")
arr7 = np.ones((10, 10), dtype=int)
arr7[1:-1, 1:-1] = 0
print(arr7)

print("\nПункт 8: 5x5, главная диагональ = 1..5")
arr8 = np.zeros((5, 5), dtype=int)
np.fill_diagonal(arr8, [1, 2, 3, 4, 5])
print(arr8)

print("\nПункт 9: 4x4 шахматный порядок с нулями на главной диагонали")
arr9 = np.indices((4, 4)).sum(axis=0) % 2
print(arr9)

print("\nПункт 10: все дни марта 2017 года")
march_2017 = np.arange('2017-03-01', '2017-04-01', dtype='datetime64[D]')
print(march_2017)
print("Тип данных:", march_2017.dtype)
```

```python
print("Пункт 1: общие элементы двух массивов")
a1 = np.array([0, 10, 20, 40, 60])
a2 = np.array([10, 30, 40])
print(np.intersect1d(a1, a2))

print("\nПункт 2: уникальные элементы")
src2a = np.array([10, 10, 20, 20, 30, 30])
src2b = np.array([[1, 1], [2, 3]])
print(np.unique(src2a))
print(np.unique(src2b))

print("\nПункт 3: уникальные элементы и их частоты")
src3 = np.array([10, 10, 20, 10, 20, 20, 20, 30, 30, 50, 40, 40])
values, counts = np.unique(src3, return_counts=True)
print(values)
print(counts)

print("\nПункт 4: повторение массива")
src4 = np.array([1, 2, 3, 4])
for k in (1, 2, 3):
    print(k, "повторение:", np.tile(src4, k))

print("\nПункт 5: удаление nan")
src5a = np.array([200., 300., np.nan, np.nan, np.nan, 700.])
print(src5a[~np.isnan(src5a)])

src5b = np.array([[1., 2., 3.],
                  [np.nan, 0., np.nan],
                  [6., 7., np.nan]])
print(src5b[~np.isnan(src5b)])

print("\nПункт 6: k наименьших значений")
src6 = np.array([1., 7., 8., 2., 0.1, 3., 15., 2.5])
print(np.sort(src6)[:4])

print("\nПункт 7: ближайший к числу элемент")
src7 = np.array([0.5, 1.8, 2.1, 3.5, 4.87, 5.13, 6.49])
target = 3.09066280756759
idx = np.abs(src7 - target).argmin()
print(src7[idx])

print("\nПункт 8: поэлементное объединение строк")
a8 = np.array(['Python', 'PHP'])
b8 = np.array(['Java', 'C++'])
print(np.char.add(np.char.add(a8, ' '), b8))

print("\nПункт 9: частота буквы 'P'")
src9 = np.array(['Python', 'PHP', 'JS', 'examples', 'html'])
print(np.char.count(src9, 'P'))

print("\nПункт 10: корни полиномов")
print("a:", np.roots([1, -4, 7]))
print("b:", np.roots([1, -11, 9, 11, -10]))
```

```python
import numpy as np

print("Решение системы линейных уравнений 2x2")
print("Введите коэффициенты системы:")
a11 = float(input("a11 = "))
a12 = float(input("a12 = "))
b1  = float(input("b1  = "))
a21 = float(input("a21 = "))
a22 = float(input("a22 = "))
b2  = float(input("b2  = "))

# Матрица коэффициентов и вектор правой части
A = np.array([[a11, a12],
              [a21, a22]])
b = np.array([b1, b2])

#определитель
det = np.linalg.det(A)

# Проверяем вырожденность
if np.isclose(det, 0):
    print("\nОпределитель равен 0.")
    print("Система не имеет решения")
else:
    # Решение системы
    solution = np.linalg.solve(A, b)
    print("\nРешение системы:")
    print(f"x = {solution[0]:.4f}")
    print(f"y = {solution[1]:.4f}")
```
